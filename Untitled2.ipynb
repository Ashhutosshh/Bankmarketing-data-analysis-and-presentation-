```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler

# 1. Load the dataset
df = pd.read_csv('Test.csv')

# 2. Drop the useless 'ID' column as it doesn't help in grouping people
df = df.drop('ID', axis=1)

# 3. Handle missing values
# For numerical columns, fill blanks with the median (middle value)
num_cols = ['Age', 'Work_Experience', 'Family_Size']
for col in num_cols:
    df[col] = df[col].fillna(df[col].median())
# For categorical (text) columns, fill blanks with the mode (most common value)
cat_cols = ['Gender', 'Ever_Married', 'Graduated', 'Profession', 'Spending_Score', 'Var_1']
for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])

# 4. Encode text to numbers
# Converts categories like 'Male'/'Female' to 0/1
le = LabelEncoder()
for col in cat_cols:
    df[col] = le.fit_transform(df[col])

# 5. Standardize the data
# Distance-based algorithms like K-Means need all numbers on a similar scale
scaler = StandardScaler()
df_scaled = pd.DataFrame(scaler.fit_transform(df), columns=df.columns)

# Save the cleaned data to a new file
df_scaled.to_csv('Test_preprocessed.csv', index=False)
print("Data preprocessed and saved successfully!")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# 1. Load the preprocessed dataset
df_scaled = pd.read_csv('Test_preprocessed.csv')

# 2. Calculate WCSS (Within-Cluster Sum of Squares) for K=1 to 10
wcss = []
k_values = range(1, 11)

for k in k_values:
    # Initialize K-Means asking for 'k' clusters
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    # Fit the model to our data
    kmeans.fit(df_scaled)
    # kmeans.inertia_ gives us the WCSS (error) value
    wcss.append(kmeans.inertia_)

# 3. Plot the Elbow Curve
plt.figure(figsize=(8, 5))
plt.plot(k_values, wcss, marker='o', linestyle='--', color='b')
plt.title('Elbow Method for Optimal K')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('WCSS (Error / Variance)')
plt.xticks(k_values)
plt.grid(True)

# Show the plot
plt.show()
```

```python
from sklearn.metrics import silhouette_score

# Assuming we already have our df_scaled and kmeans model with 4 clusters
cluster_labels = kmeans.fit_predict(df_scaled)

# Calculate the score
score = silhouette_score(df_scaled, cluster_labels)
print(f"Silhouette Score: {score}")
```
